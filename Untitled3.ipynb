```python
import random
options =("rock","paper","scissor")
playing=True
while playing:
    player=None
    computer=random.choice(options)

    while player not in options:
        player=input("enter any one (rock,paper,scissor): ").lower()
    print(f"player:{player}")
    print(f"computer:{computer}")
    if player==computer:
        print("its tie")
    elif player=="rock" and computer=="scissor":
            print("you win!")
    elif player =="paper" and computer =="rock":
            print("you win!")
    elif player=="scissor" and computer=="paper":
            print("you win!")
    else:
            print("you lose!")
    if not input("Do you want to play (y/n) :")=="y":
             break
print("thank you for playing ,see u again")
```
